## Criterio empleado

En clase vimos que el umbral epidémico para una red heterogénea es:

$$\frac{\beta}{\gamma} > \frac{\langle k\rangle}{\langle k^2\rangle-\langle k\rangle}$$

Equivalentemente, se define el número reproductivo de red
$$\mathcal R_{red}=\frac{\beta}{\gamma}\frac{\langle k^2\rangle-\langle k\rangle}{\langle k\rangle},$$
y hay propagación si $\mathcal R_{red}>1$. El valor crítico de transmisión es
$$\beta_c=\gamma\frac{\langle k\rangle}{\langle k^2\rangle-\langle k\rangle}.$$


## Task 2.1

Se tienen $\beta=0.08\;\text{día}^{-1}$, $\gamma=0.05\;\text{día}^{-1}$, $\langle k\rangle=8.4$ y $\langle k^2\rangle=210.6$.

### a. Umbral y propagación sin intervención

$$\frac{\beta}{\gamma}=\frac{0.08}{0.05}=1.6$$
$$\frac{\langle k\rangle}{\langle k^2\rangle-\langle k\rangle}=\frac{8.4}{210.6-8.4}=\frac{8.4}{202.2}=0.04154$$

Como $1.6>0.04154$, se satisface la condición de epidemia. De forma equivalente:
$$\mathcal R_{red}=1.6\left(\frac{202.2}{8.4}\right)=38.51>1,$$
$$\beta_c=0.05\left(\frac{8.4}{202.2}\right)=0.00208\;\text{día}^{-1}. $$

**Conclusión:** sí, el brote puede propagarse; $\beta=0.08$ está muy por encima de $\beta_c$.

### b. Evaluación de estrategias

**Estrategia A: mascarillas.** Reducir $\beta$ en 40% da $\beta_A=0.08(1-0.40)=0.048$. Los momentos no cambian, por lo que:
$$\mathcal R_{red,A}=\frac{0.048}{0.05}\left(\frac{202.2}{8.4}\right)=23.11>1.$$
También $\beta_A=0.048>\beta_c=0.00208$. Por tanto, **A no es suficiente** para bajar del umbral.

**Estrategia B: vacunación aleatoria.** El enunciado establece una reducción de 30% en el primer momento y de 51% en el segundo:
$$\langle k\rangle_B=8.4(0.70)=5.88,$$
$$\langle k^2\rangle_B=210.6(0.49)=103.194.$$
Entonces
$$\mathcal R_{red,B}=\frac{0.08}{0.05}\left(\frac{103.194-5.88}{5.88}\right)=26.48>1,$$
y $\beta_{c,B}=0.05(5.88)/(103.194-5.88)=0.00302\;\text{día}^{-1}<0.08$. Por tanto, **B tampoco es suficiente**. Aunque reduce el riesgo respecto al caso inicial, no alcanza la reducción necesaria.

### c. Consistencia teórica y alternativa

Sea $p=0.70$ la fracción de nodos (y, aproximadamente, de vecinos por nodo) que permanece tras la vacunación aleatoria. El grado remanente de un nodo de grado original $k$ se adelgaza binomialmente: $K'\mid k\sim\mathrm{Binomial}(k,p)$. Por ello,
$$\langle K'\rangle=p\langle k\rangle,$$
$$\langle {K'}^2\rangle=p^2\langle k^2\rangle+p(1-p)\langle k\rangle.$$

El término dominante para una red heterogénea, $p^2\langle k^2\rangle=0.49\langle k^2\rangle$, predice aproximadamente una reducción de 51%; el término adicional $p(1-p)\langle k\rangle$ es pequeño frente a $\langle k^2\rangle$ en estos datos. Así, el 51% dado es la aproximación dominante esperada bajo eliminación aleatoria. Sin embargo, esta intervención conserva proporcionalmente los hubs, que dominan $\langle k^2\rangle$, y por eso no elimina bien el riesgo en redes heterogéneas.

La alternativa más eficiente es la **vacunación dirigida por grado**: seleccionar primero los nodos de mayor $k$. Eliminar un nodo de grado $k$ aporta una reducción proporcional a $k$ en $\langle k\rangle$, pero a $k^2$ en $\langle k^2\rangle$; por eso retirar hubs reduce de manera desproporcionada el factor que controla $\mathcal R_{red}$.


## Task 2.2

### a. Municipio X: distribución Poisson

Para una Poisson con media $\lambda=\langle k\rangle=12$, se cumple
$$\langle k^2\rangle=\lambda^2+\lambda=12^2+12=156.$$
Por consiguiente,
$$\mathcal R_{red,X}=\frac{0.08}{0.05}\frac{156-12}{12}=1.6(12)=19.2,$$
$$\beta_{c,X}=0.05\frac{12}{144}=0.00417\;\text{día}^{-1}.$$

El SIR clásico da
$$R_{0,\mathrm{clásico}}=\frac{\beta\langle k\rangle}{\gamma}=\frac{0.08(12)}{0.05}=19.2.$$
Son iguales en este caso porque para Poisson $\langle k^2\rangle-\langle k\rangle=\lambda^2$, y por tanto el exceso de grado medio es $\lambda$. No son idénticos en general: la fórmula de red incorpora la heterogeneidad mediante $\langle k^2\rangle$, mientras que el SIR clásico solo usa el promedio.

### b. Municipio Y: ley de potencias

Para $P(k)\sim k^{-\alpha}$, con $\alpha=2.4$, $k_{min}=1$ y $k_{max}=800$, la aproximación indicada para $2<\alpha<3$ es
$$\langle k^2\rangle\approx\frac{\alpha-1}{3-\alpha}k_{min}^{\alpha-1}k_{max}^{3-\alpha}.$$
Sustituyendo:
$$\langle k^2\rangle\approx\frac{1.4}{0.6}(1)^{1.4}(800)^{0.6}=128.77.$$

Para el primer momento de la misma aproximación,
$$\langle k\rangle\approx\frac{\alpha-1}{\alpha-2}k_{min}=\frac{1.4}{0.4}(1)=3.5.$$
Así,
$$\mathcal R_{red,Y}=1.6\left(\frac{128.77-3.5}{3.5}\right)=57.27>1,$$
$$\beta_{c,Y}=0.05\left(\frac{3.5}{128.77-3.5}\right)=0.00140\;\text{día}^{-1}. $$

**Conclusión:** no puede contenerse con $\beta=0.08$ y $\gamma=0.05$ sin una intervención adicional. Su umbral es todavía menor que en X debido al gran segundo momento producido por los hubs. (Si se retiene el término finito $-k_{min}^{3-\alpha}$, se obtiene $\langle k^2\rangle\approx126.44$; la conclusión no cambia.)

### c. Vacunar 20%: comparación basada en momentos

| Municipio | Vacunación aleatoria (20%) | Vacunación dirigida a mayor grado (20%) | Consecuencia sobre el umbral |
|:--|:--|:--|:--|
| **X (Poisson)** | Con $p=0.8$, $\langle k\rangle'=p\langle k\rangle=9.6$ y $\langle {k'}^2\rangle=p^2(156)+p(1-p)(12)=101.76$. La dispersión de grados es estrecha. | Los grados están concentrados alrededor de 12, así que los nodos de mayor grado no aportan una fracción extraordinaria de $\langle k^2\rangle$. | Ambas campañas reducen los dos momentos; la dirigida es mejor, pero su ventaja es moderada porque no hay hubs. La aleatoria es una opción relativamente efectiva en una Poisson. |
| **Y (ley de potencias)** | Con $p=0.8$, el aporte dominante al segundo momento se escala aproximadamente como $p^2\langle k^2\rangle=0.64\langle k^2\rangle$ (más el término binomial $p(1-p)\langle k\rangle$). Aún sobreviven hubs y el segundo momento sigue alto. | Al retirar primero los mayores $k$, se recorta la cola superior: cada hub removido sustrae una contribución de orden $k^2$ a $\langle k^2\rangle$. Esto baja mucho más $\langle k^2\rangle$ que $\langle k\rangle$. | La dirigida es claramente superior: disminuye de forma mucho más intensa el cociente $(\langle k^2\rangle-\langle k\rangle)/\langle k\rangle$, eleva $\beta_c$ y ataca el origen del umbral bajo. |

En suma, la comparación no depende solo de cuántas personas se vacunan: depende de qué parte de los momentos de grado se elimina. En X, los grados similares hacen que la vacunación aleatoria reduzca los momentos de manera suficientemente representativa. En Y, la masa de $\langle k^2\rangle$ está concentrada en la cola, por lo que la misma cobertura debe asignarse a los grados altos para tener el mayor efecto epidemiológico.


```python
# Verificación numérica de los valores reportados
beta, gamma = 0.08, 0.05

def R_red(beta, gamma, k1, k2):
    return beta / gamma * (k2 - k1) / k1

def beta_critica(gamma, k1, k2):
    return gamma * k1 / (k2 - k1)

k1, k2 = 8.4, 210.6
k1_B, k2_B = 0.70 * k1, 0.49 * k2
alpha, kmin, kmax = 2.4, 1, 800
k2_Y = (alpha - 1) / (3 - alpha) * kmin**(alpha - 1) * kmax**(3 - alpha)
k1_Y = (alpha - 1) / (alpha - 2) * kmin

print(f'Task 2.1: R_red = {R_red(beta, gamma, k1, k2):.2f}; beta_c = {beta_critica(gamma, k1, k2):.5f}')
print(f'Estrategia A: R_red = {R_red(0.60 * beta, gamma, k1, k2):.2f}')
print(f'Estrategia B: R_red = {R_red(beta, gamma, k1_B, k2_B):.2f}; beta_c = {beta_critica(gamma, k1_B, k2_B):.5f}')
print(f'Municipio X: R_red = {R_red(beta, gamma, 12, 156):.2f}; beta_c = {beta_critica(gamma, 12, 156):.5f}')
print(f'Municipio Y: <k^2> = {k2_Y:.2f}; R_red = {R_red(beta, gamma, k1_Y, k2_Y):.2f}; beta_c = {beta_critica(gamma, k1_Y, k2_Y):.5f}')
```
